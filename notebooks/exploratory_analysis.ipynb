# Exploratory Analysis & Disturbance Characterization

This notebook demonstrates exploratory analysis, spectral evaluation, damping ratio estimation, and multi-channel PMU source assessment using the `grid_disturbance` package.

```python
import numpy as np
import matplotlib.pyplot as plt
from grid_disturbance.synthetic.generator import SyntheticSignalGenerator
from grid_disturbance.signals.spectral import analyze_spectrum
from grid_disturbance.signals.envelopes import extract_hilbert_envelope, fit_exponential_envelope
from grid_disturbance.analysis.diagnosis import DisturbanceDiagnosticEngine
```

## 1. Generating a Natural Ring-Down Disturbance
We generate a synthetic ring-down signal with an 8.0% damping ratio at 0.65 Hz.

```python
gen = SyntheticSignalGenerator(random_seed=42)
sig = gen.generate_damped_ringdown(
    frequency=0.65, damping_ratio=0.08, amplitude=1.5, sampling_rate=30.0
)
print(f"Samples: {len(sig.data)}, True Damping: {sig.ground_truth.damping_ratio * 100:.1f}%")
```

## 2. Spectral Analysis
Extract power spectral density, peak frequency, and spectral concentration.

```python
spec = analyze_spectrum(sig.data, sampling_rate=30.0)
print(f"Estimated Dominant Frequency: {spec.dominant_frequency:.2f} Hz")
print(f"Spectral Concentration: {spec.spectral_concentration:.2f}")
```

## 3. End-to-End Disturbance Diagnosis
Execute the unified diagnostic engine to discriminate between forced and natural mechanisms.

```python
engine = DisturbanceDiagnosticEngine()
diag = engine.diagnose_signal(sig.data, sampling_rate=30.0)
print(f"Predicted Class: {diag.disturbance_type}")
print(f"Confidence: {diag.confidence * 100:.1f}%")
print(f"Summary: {diag.summary}")
```
